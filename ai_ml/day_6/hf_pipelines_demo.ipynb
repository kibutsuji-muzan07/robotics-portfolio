```python
# Day B-6: Hugging Face Transformers — pipeline() API
from transformers import pipeline
```

```python
# ── 1. Sentiment Classification ───────────────────────────────────────────
print("=" * 60)
print("SENTIMENT CLASSIFIER")
print("=" * 60)

classifier = pipeline("text-classification", model="distilbert-base-uncased-finetuned-sst-2-english")
```

```python
sentences = [
    "I love building robots and autonomous systems.",
    "The deployment pipeline kept failing all night.",
    "Machine learning is transforming every industry.",
    "I have no idea what I am doing with this code.",
    "The model achieved 97% accuracy on the test set.",
    "This is the worst API documentation I have ever read.",
    "Gradient descent finally clicked for me today.",
    "The AWS costs are completely out of control.",
    "I genuinely enjoy learning about transformers.",
    "My pull request has been stuck in review for two weeks.",
]

for s in sentences:
    result = classifier(s)[0]
    print(f"{result['label']:<10} {result['score']:.4f}  |  {s}")
```

```python
# ── 2. Summarization ───────────────────────────────────────────────────────
print("\n" + "=" * 60)
print("SUMMARIZER")
print("=" * 60)

summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

long_text = """
Retrieval-Augmented Generation (RAG) is a technique that combines
information retrieval with large language model generation. Instead of
relying solely on the knowledge encoded in the model's weights during
training, RAG systems first retrieve relevant documents from an external
knowledge base, then pass those documents along with the original query
to the language model as context. This allows the model to generate
answers grounded in up-to-date, domain-specific information that may not
have been present in its training data. The retrieval step typically uses
dense vector search with embeddings and cosine similarity to find the
most semantically relevant passages. The generation step uses the
retrieved context as a prompt prefix so the model can produce accurate,
sourced answers rather than hallucinating from memory alone.
"""

summary = summarizer(long_text, max_length=80, min_length=30, do_sample=False)
print("Original length:", len(long_text.split()), "words")
print("Summary:", summary[0]['summary_text'])
```

```python
# ── 3. Print token IDs for one sentence ───────────────────────────────────
print("\n" + "=" * 60)
print("TOKENIZATION DEMO")
print("=" * 60)

from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
sample = "Gradient descent finally clicked for me today."
tokens = tokenizer(sample, return_tensors=None)
print("Text   :", sample)
print("IDs    :", tokens['input_ids'])
print("Tokens :", tokenizer.convert_ids_to_tokens(tokens['input_ids']))
```
