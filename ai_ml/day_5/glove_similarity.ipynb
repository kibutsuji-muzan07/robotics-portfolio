```python
# Day B-5: Word Embeddings + Cosine Similarity with GloVe
import numpy as np

def load_glove(filepath):
    embeddings = {}
    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            parts = line.split()
            word = parts[0]
            vector = np.array(parts[1:], dtype=np.float32)
            embeddings[word] = vector
    print(f"Loaded {len(embeddings)} word embeddings (dim={vector.shape[0]})")
    return embeddings

glove = load_glove("glove.6B.50d.txt")

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

def most_similar(word, embeddings, top_n=5):
    if word not in embeddings:
        print(f"'{word}' not in vocabulary")
        return []
    query_vec = embeddings[word]
    scores = [(w, cosine_similarity(query_vec, vec))
              for w, vec in embeddings.items() if w != word]
    scores.sort(key=lambda x: x[1], reverse=True)
    return scores[:top_n]

for qw in ["robot", "drone", "learn"]:
    print(f"\nTop 5 most similar to '{qw}':")
    for word, score in most_similar(qw, glove):
        print(f"  {word:<20} {score:.4f}")

# Word arithmetic
if all(w in glove for w in ["king", "man", "woman"]):
    result_vec = glove["king"] - glove["man"] + glove["woman"]
    scores = [(w, cosine_similarity(result_vec, vec))
              for w, vec in glove.items() if w not in ["king","man","woman"]]
    scores.sort(key=lambda x: x[1], reverse=True)
    print(f"\nking - man + woman = '{scores[0][0]}' ({scores[0][1]:.4f})")
```
